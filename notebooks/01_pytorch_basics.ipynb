```python
import torch

t = torch.tensor([[1, 2, 3],
                  [4, 5, 6]])
print(t)
print(t.shape)
print(t.dtype)
```

```python
a = torch.tensor([[1., 2.],
                  [3., 4.]])
b = torch.tensor([[10., 0.],
                  [0., 10.]])
print(a * b)     
print(a @ b)     
```

```python
x = torch.ones(2, 3)
y = torch.ones(2, 3)
print(x @ y.T)
```

```python
a = torch.tensor(2.0, requires_grad=True)
b = torch.tensor(-3.0, requires_grad=True)
c = torch.tensor(10.0, requires_grad=True)

d = a * b + c
e = d * a
e.backward()

print(e)
print(a.grad, b.grad, c.grad)
```
