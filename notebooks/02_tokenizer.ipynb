```python
import sentencepiece as spm

sp = spm.SentencePieceProcessor(model_file="../data/raw/kikamba-corpus/tokenizer_models/kam_bpe_8k.model")
print(sp.get_piece_size())
print(sp.encode("Ũũ nĩ mũtwe mũnene.", out_type=str))
print(sp.encode("Ũũ nĩ mũtwe mũnene."))
```

```python
ids = sp.encode("Ũũ nĩ mũtwe mũnene.")
print(sp.decode(ids))
print(sp.decode(ids) == "Ũũ nĩ mũtwe mũnene.")
```

```python
lines = open("../data/splits/val.txt", encoding="utf-8").read().split("\n")
lines = [l for l in lines if l.strip()]

bad = 0
for line in lines:
    if sp.decode(sp.encode(line)) != line:
        bad += 1

print(len(lines), bad)
```

```python
chars = 0
tokens = 0
for line in lines:
    chars += len(line)
    tokens += len(sp.encode(line))
print(chars / tokens)
```

```python
import numpy as np

def encode_file(name):
    text = open(f"../data/splits/{name}.txt", encoding="utf-8").read()
    ids = []
    for line in text.split("\n"):
        if line.strip():
            ids.extend(sp.encode(line))
    arr = np.array(ids, dtype=np.uint16)
    arr.tofile(f"../data/splits/{name}_ids.bin")
    print(name, len(arr))

encode_file("train")
encode_file("val")
```
