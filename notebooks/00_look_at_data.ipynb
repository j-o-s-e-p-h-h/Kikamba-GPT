```python
import json

path = "../data/raw/kikamba-corpus/data/monolingual/train.jsonl"

records = []
with open(path, encoding="utf-8") as f:
    for line in f:
        records.append(json.loads(line))

print(len(records))
```

```python
for r in records[:3]:
    print(r["source"], "|", r["license"], "|", r["genre"])
    print(r["text"][:200])
    print("-----")
```

```python
print(len(records))
total_words = 0
total_chars = 0
for r in records:
    total_words += len(r["text"].split())
    total_chars += len(r["text"])

print(total_words)
print(total_chars)
```

```python
total = 0    
words = 0
for r in records:
    if 'NC' in r["license"] or 'ND' in r["license"]:
        total += 1
        words += len(r["text"].split())

print(total)
print(words)
```

```python
kept = []
for r in records:
    if not'NC' in r["license"] and not'ND' in r["license"]:
        kept.append(r)
print(len(kept))
```

```python
import random 
random.Random(1337).shuffle(kept)
n_val = int(0.05 * len(kept))
val = kept[:n_val]
train = kept[n_val:]
print(len(train))
print(len(val))
```

```python
train_texts = set()
for r in train:
    train_texts.add(r["text"])
print(len(train_texts))
```

```python
val_texts = set()
for r in val:
    val_texts.add(r["text"])
print(len(train_texts))
print(len(train_texts & val_texts))
```

```python
with open("../data/splits/train.txt", "w", encoding="utf-8") as f:
    for r in train:
        f.write(r["text"] + "\n\n")
```

```python
with open("../data/splits/val.txt", "w", encoding="utf-8") as f:
    for r in val:
        f.write(r["text"] + "\n\n")
print(len(open("../data/splits/val.txt", encoding="utf-8").read()))
```

```python
sample = random.Random(1).sample(kept, 20)
for i, r in enumerate(sample, 1):
    print(i, "|", r["source"])
    print(r["text"][:300])
    print("-----")
```
