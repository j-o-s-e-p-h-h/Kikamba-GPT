```python
import json

path = "../data/raw/kikamba-corpus/data/monolingual/train.jsonl"

records = []
with open(path, encoding="utf-8") as f:
    for line in f:
        records.append(json.loads(line))

print(len(records))
```

```python
for r in records[:3]:
    print(r["source"], "|", r["license"], "|", r["genre"])
    print(r["text"][:200])
    print("-----")
```
